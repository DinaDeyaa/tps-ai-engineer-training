## 3. Simple calculator

```python
number1 = 12
number2 = 4
operator = "+"

if operator == "+":
    print(number1 + number2)
elif operator == "-":
    print(number1 - number2)
elif operator == "*":
    print(number1 * number2)
elif operator == "/":
    print(number1 / number2)
else:
    print("Invalid operator")
```

## 9. Count vowels in a string

```python
text = "Hello Python"
vowel_count = 0

for letter in text.lower():
    if letter in "aeiou":
        vowel_count += 1

print("Vowels:", vowel_count)
```

## 11. Remove duplicates from a list

```python
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = []

for number in numbers:
    if number not in unique_numbers:
        unique_numbers.append(number)

print(unique_numbers)
```

## 12. Filter even numbers from a list

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)

print(even_numbers)
```

## 14. Word frequency counter

```python
text = "python is fun python is easy"
word_counts = {}

for word in text.lower().split():
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print(word_counts)
```

## 16. Factorial function

```python
def factorial(number):
    result = 1
    for i in range(1, number + 1):
        result *= i
    return result


number = 5
print("Factorial:", factorial(number))
```

## 17. Palindrome checker

```python
def is_palindrome(word):
    reversed_word = ""
    for letter in word:
        reversed_word = letter + reversed_word
    return word == reversed_word


word = "level"
print("Palindrome:", is_palindrome(word))
```

## 19. Count lines in a text file

```python
def count_lines(filename):
    count = 0
    with open(filename, "r") as file:
        for line in file:
            count += 1
    return count


print("Lines:", count_lines("practice.txt"))
```

## 20. Safe division using try/except

```python
def safe_division(number1, number2):
    try:
        return number1 / number2
    except ZeroDivisionError:
        return "Cannot divide by zero"
    except TypeError:
        return "Invalid inputs"


print(safe_division(10, 2))
print(safe_division(10, 0))
print(safe_division(10, "hello"))
```
